```python
# YouTube channel link
# Extract video links from that url
# Get the IDs of each video
# Get the transcripts of those videos
# Use gemini api to get summaries from those videos
# Done.

import yt_dlp
from yt_dlp.utils import DateRange  # for handling internal date bounds
from yt_dlp.utils import match_filter_func # for filtering with video length
from yt_dlp.utils import DownloadCancelled
```

```python
channel_url = "https://www.youtube.com/@sigchi/videos"

MIN_SECONDS = 120   # 2 minutes
MAX_SECONDS = 1200  # 20 minutes

def custom_filter(info_dict, incomplete):
    # 1. Fetch metadata values safely
    duration = info_dict.get('duration')
    upload_date = info_dict.get('upload_date')  # Format: "YYYYMMDD"

    # 2. Check the date constraint (Only 2026)
    if upload_date:
        # Stop fetching completely if we encounter a video older than Jan 1, 2026
        if upload_date < '20260101':
            raise DownloadCancelled("Reached videos older than 2026. Stopping.")
        
        # Skip videos ahead of 2026
        if upload_date > '20261231':
            return "Video is newer than 2026"

    # 3. Check the duration constraint
    if duration is not None:
        if duration < MIN_SECONDS or duration > MAX_SECONDS:
            return f"Skipped: Duration {duration}s out of bounds"

    return None  # Return None to accept and process the video

```

```python
ydl_opts = {
    'extract_flat': True,       # Fast extraction (doesn't load full video web pages)
    'quiet': False,             # Display logs to console
    'playlistend': 30,          # Caps the search at 30 videos
    'match_filter': custom_filter,  # Assigns our custom filter logic function
}
```

```python
print("Fetching video links using yt-dlp... please wait.")

with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    # Extract playlist/channel information
    result = ydl.extract_info(channel_url, download=False)

    if 'entries' in result:
        with open("../data/youtube_links.txt", "w") as file:
            for entry in result['entries']:
                if entry and 'id' in entry:
                    video_url = f"https://www.youtube.com/watch?v={entry['id']}"
                    file.write(video_url + "\n")
                    print(entry['id'])

print("\nDone! Links saved to youtube_links.txt")
```

```python
def get_video_info(url):
    # Get YouTube video metadata.
    options = {
        "quiet": True,
        "no_warnings": True,
        "skip_download": True,
    }

    with yt_dlp.YoutubeDL(options) as ydl:
        info = ydl.extract_info(url, download=False)

    return {
        "id": info["id"],
        "title": info["title"],
        "url": url,
    }
```

```python
# Read URLs
with open("../data/youtube_links.txt", "r", encoding="utf-8") as f:
    urls = [line.strip() for line in f if line.strip()]

videos = {}

for url in urls:
    try:
        info = get_video_info(url)

        videos[info["id"]] = {
            "title": info["title"],
            "url": info["url"],
            "transcript": ""
        }

        print(f"{info['id']}: {info['title']}")

    except Exception as e:
        print(f"Failed: {url}")
        print(f"  {e}")
```

```python
import json
from youtube_transcript_api import YouTubeTranscriptApi
from youtube_transcript_api.formatters import TextFormatter


ytt_api = YouTubeTranscriptApi()
formatter = TextFormatter()
```

```python
for video_id, video in videos.items():
    try:
        transcript = ytt_api.fetch(video_id, languages=["en"])
        text = formatter.format_transcript(transcript)

        videos[video_id]["transcript"] = text
        with open("youtube_videos.json", "w", encoding="utf-8") as f:
            json.dump(videos, f, indent=2, ensure_ascii=False)

        print(f"{video['title']}")

    except Exception as e:
        print(f"{video['title']}")
        print(f"Error: {e}")
```

```python
# Now I got the transcripts the next step is getting summaries using AI.

import os
from dotenv import load_dotenv
from google import genai

load_dotenv()

client = genai.Client()

```

```python
def generate_summary(video_id, title, transcript):

    prompt = f"""
You are analyzing a research video from the ACM CHI Conference.

Video title:
{title}

Analyze the transcript below and provide a concise but informative summary.

Use the following structure:

### What is it about?
Explain the research problem, proposed system/interaction, and how it works.

### Key Innovation
Explain what is novel or interesting about the HCI technique or system.

### Potential Applications
Explain who could use this system and what it could be useful for.

### Interesting Points
List 3-5 particularly interesting ideas, findings, or design decisions.

### Limitations
Mention any limitations, challenges, or concerns that can reasonably be identified from the transcript.

Do not invent information that is not supported by the transcript.

Transcript:
{transcript}
"""

    max_retries = 3

    for attempt in range(max_retries):

        try:
            response = client.models.generate_content(
                model="gemini-3.1-flash-lite",
                contents=prompt
            )

            return response.text

        except Exception as e:

            print(
                f"Attempt {attempt + 1}/{max_retries} failed: {e}"
            )

            if attempt < max_retries - 1:
                wait_time = 15 * (attempt + 1)

                print(
                    f"Waiting {wait_time} seconds before retry..."
                )

                time.sleep(wait_time)

            else:
                print(f"Failed permanently: {video_id}")
                return ""
```

```python
import time

for i, (video_id, video) in enumerate(videos.items(), start=1):

    # Skip videos that already have a summary
    if video.get("summary"):
        print(f"[{i}/{len(videos)}] Already done: {video['title']}")
        continue

    # Skip videos without transcripts
    if not video.get("transcript"):
        print(f"[{i}/{len(videos)}] No transcript: {video['title']}")
        continue

    print(f"\n[{i}/{len(videos)}] Processing:")
    print(video["title"])

    try:
        summary = generate_summary(
            video_id,
            video["title"],
            video["transcript"]
        )

        video["summary"] = summary

        print("Summary generated")

    except Exception as e:
        print(f"✗ Error: {e}")
        continue

    # Save immediately after every successful request
    with open("../data/youtube_videos_analyzed.json", "w", encoding="utf-8") as f:
        json.dump(
            videos,
            f,
            indent=2,
            ensure_ascii=False
        )

    # Wait before next API request
    if i < len(videos):
        print("Waiting 5 seconds...")
        time.sleep(5)
```
